# Week 12: Implement & Analyze Unsupervised Learning Techniques for Pattern Discovery
**Techniques:** K-Means Clustering and Hierarchical Clustering

**Application:** Customer Segmentation

## Create Dataset
Synthetic customer data (Annual Income, Spending Score) generated with `make_blobs`.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=100,
    centers=3,
    cluster_std=1.5,
    random_state=42
)

print("First 5 customers [Annual Income, Spending Score]:")
print(X[:5])
```

## 1. K-Means Clustering

```python
print("----- K-Means Clustering -----")

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

kmeans_labels = kmeans.labels_
centers = kmeans.cluster_centers_

print("Cluster Labels:")
print(kmeans_labels)
print("\nCluster Centers:")
print(centers)
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels)
plt.scatter(centers[:, 0], centers[:, 1], marker='X', s=200, color='red')
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

## 2. Hierarchical Clustering

```python
print("----- Hierarchical Clustering -----")

linked = linkage(X, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```
