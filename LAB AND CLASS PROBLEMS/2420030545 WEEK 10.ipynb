# Week 10: Implement & Analyze Regression Models for Predictive Analytics
**Application:** House Price Prediction (Linear Regression) and House Purchase Classification (Logistic Regression)

## 1. Linear Regression – House Price Prediction

```python
import numpy as np
from sklearn.linear_model import LinearRegression, LogisticRegression

# Independent variable: house size (sq.ft)
X_linear = np.array([[1000], [1200], [1500], [1800], [2000]])

# Dependent variable: house price (lakhs)
y_linear = np.array([50, 55, 65, 80, 90])

print("Training House Sizes:")
print(X_linear.flatten())
print("\nTraining House Prices:")
print(y_linear)
```

```python
# Create and train the Linear Regression model
linear_model = LinearRegression()
linear_model.fit(X_linear, y_linear)

print("Slope (m):", linear_model.coef_[0])
print("Intercept (c):", linear_model.intercept_)
```

```python
# Predict price for a new 1700 sq.ft house
new_house_size = np.array([[1700]])
predicted_price = linear_model.predict(new_house_size)

print("Predicted Price for 1700 sq.ft house (lakhs):")
print(predicted_price[0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_linear, y_linear, label="Training data")
plt.plot(X_linear, linear_model.predict(X_linear), color="red", label="Best-fit line")
plt.scatter(new_house_size, predicted_price, color="green", s=80, label="Prediction (1700 sq.ft)")
plt.xlabel("House Size (sq.ft)")
plt.ylabel("Price (Lakhs)")
plt.title("Linear Regression: House Price Prediction")
plt.legend()
plt.show()
```

## 2. Logistic Regression – House Purchase Classification

```python
# Independent variable: house price (lakhs)
X_logistic = np.array([[40], [50], [60], [70], [80], [90]])

# Dependent variable: 0 = Not Purchased, 1 = Purchased
y_logistic = np.array([0, 0, 0, 1, 1, 1])

logistic_model = LogisticRegression()
logistic_model.fit(X_logistic, y_logistic)

print("Training House Prices:")
print(X_logistic.flatten())
print("\nPurchase Labels:")
print(y_logistic)
```

```python
# Predict purchase decision for a 75 lakh house
test_price = np.array([[75]])
purchase_prediction = logistic_model.predict(test_price)
purchase_probability = logistic_model.predict_proba(test_price)

print("Test House Price:", test_price[0][0])
print("\nPurchase Prediction:")
print(purchase_prediction[0])
print("\nPrediction Probabilities [Not Purchased, Purchased]:")
print(purchase_probability)
```

```python
prices = np.linspace(30, 100, 200).reshape(-1, 1)
probs = logistic_model.predict_proba(prices)[:, 1]

plt.scatter(X_logistic, y_logistic, label="Training data")
plt.plot(prices, probs, color="red", label="Sigmoid curve P(y=1)")
plt.axhline(0.5, linestyle="--", color="gray")
plt.xlabel("House Price (Lakhs)")
plt.ylabel("Probability of Purchase")
plt.title("Logistic Regression: Purchase Probability")
plt.legend()
plt.show()
```
