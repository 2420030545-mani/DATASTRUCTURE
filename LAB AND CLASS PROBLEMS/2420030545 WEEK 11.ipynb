# Week 11: Implement & Analyze Classification Algorithms using Performance Evaluation Metrics
**Application:** Student Pass/Fail Prediction

**Metrics:** Accuracy, Error Rate, Precision, Recall, F1-Score, Confusion Matrix

```python
import numpy as np
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix)

# Features: [Study Hours, Attendance Percentage]
X_train = np.array([
    [2, 50], [3, 60], [4, 65], [5, 70],
    [6, 75], [7, 80], [8, 85], [9, 90]
])

# Labels: 0 = Fail, 1 = Pass
y_train = np.array([0, 0, 0, 1, 1, 1, 1, 1])

# Test data
X_test = np.array([[3, 55], [5, 72], [8, 88], [4, 60]])
y_test = np.array([0, 1, 1, 0])
```

## 11.1 Decision Tree
Rule-based, easy to visualize, prone to overfitting.

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
```

```python
print("Actual Labels:")
print(y_test)
print("\nPredicted Labels:")
print(y_pred)
print("\nConfusion Matrix:")
print(cm)
print("\nAccuracy:")
print(accuracy)
print("\nError Rate:")
print(error_rate)
print("\nPrecision:")
print(precision)
print("\nRecall:")
print(recall)
print("\nF1-Score:")
print(f1)
```

## 11.2 K-Nearest Neighbors (KNN)
Instance-based, simple but computationally heavy for large datasets.

```python
from sklearn.neighbors import KNeighborsClassifier

# k = 3 nearest neighbors
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
```

```python
print("Actual Labels:")
print(y_test)
print("\nPredicted Labels:")
print(y_pred)
print("\nConfusion Matrix:")
print(cm)
print("\nAccuracy:")
print(accuracy)
print("\nError Rate:")
print(error_rate)
print("\nPrecision:")
print(precision)
print("\nRecall:")
print(recall)
print("\nF1-Score:")
print(f1)
```
